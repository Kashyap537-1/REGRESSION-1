#1. What is Simple Linear Regression?
- Simple Linear Regression is a statistical method used to model and quantify the relationship between two continuous variables:Independent Variable ($X$): The predictor or explanatory variable.Dependent Variable ($Y$): The response or outcome variable.It assumes a linear relationship between these variables and fits a straight line (the regression line) through the data points to predict the value of $Y$ based on given values of $X$.
#2. What are the key assumptions of Simple Linear Regression?
-   The standard assumptions required for valid inferences in Simple Linear Regression are:Linearity: The relationship between the independent and dependent variable is linear.Independence: The observations are independent of each other (no autocorrelation in errors).Homoscedasticity: The variance of the residual/error terms is constant across all levels of the independent variable.Normality: The error terms (residuals) are normally distributed.

#3. What does the coefficient m represent in the equation $Y=mX+c$?
-  The coefficient $m$ represents the slope of the regression line. It quantifies the change in the dependent variable ($Y$) for every one-unit increase in the independent variable ($X$).If $m > 0$, there is a positive relationship (as $X$ increases, $Y$ increases).If $m < 0$, there is a negative relationship (as $X$ increases, $Y$ decreases).

#4. What does the intercept c represent in the equation $Y=mX+c$?
- The intercept $c$ (often denoted as $\beta_0$) represents the y-intercept. It is the expected or predicted value of the dependent variable ($Y$) when the independent variable ($X$) is equal to zero.


#5. How do we calculate the slope m in Simple Linear Regression?The slope $m$ is calculated using the formula:$$m = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{\sum (X_i - \bar{X})^2}$$Alternatively, it can be expressed using the correlation coefficient ($r$) and the standard deviations of $X$ ($s_x$) and $Y$ ($s_y$):$$m = r \times \frac{s_y}{s_x}$$
#6. What is the purpose of the least squares method in Simple Linear Regression?-  The purpose of the Ordinary Least Squares (OLS) method is to find the optimal straight line that best fits the data points. It achieves this by minimizing the sum of the squared differences (residuals) between the actual observed values ($Y$) and the predicted values ($\hat{Y}$) generated by the regression line.
#7. How is the coefficient of determination ($R^{2}$) interpreted in Simple Linear Regression?
- $R^2$ measures the proportion of variance in the dependent variable ($Y$) that can be predicted or explained by the independent variable ($X$).An $R^2$ of 0.85, for example, means that 85% of the variation in $Y$ is explained by $X$, while the remaining 15% is due to unexplained variation or error. It ranges from 0 to 1.
#8. What is Multiple Linear Regression?
-  Multiple Linear Regression is an extension of simple linear regression used to model the linear relationship between one continuous dependent variable ($Y$) and two or more independent variables ($X_1, X_2, \dots, X_n$).Its general equation is:$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_n X_n + \epsilon$$
#9. What is the main difference between Simple and Multiple Linear Regression?-
  Simple Linear Regression: Utilizes exactly one independent variable to predict the dependent variable.Multiple Linear Regression: Utilizes two or more independent variables to predict the dependent variable simultaneously, allowing the model to control for confounding factors.
#10. What are the key assumptions of Multiple Linear Regression?
-   Multiple Linear Regression shares the same assumptions as simple linear regression, with one major addition:Linearity: Linear relationship between features and target.Independence: No autocorrelation in errors.Homoscedasticity: Constant variance of residuals.Normality: Residuals are normally distributed.No Multicollinearity: The independent variables should not be highly correlated with each other.
#11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?
-   Heteroscedasticity occurs when the variance of the residuals is not constant across all levels of the independent variables (e.g., error spreads out in a cone shape).Effects: It does not bias the regression coefficients themselves, but it invalidates the standard errors. As a result, p-values and confidence intervals become unreliable, potentially leading to incorrect conclusions about a variable's statistical significance.
#12. How can you improve a Multiple Linear Regression model with high multicollinearity?
-   To address high multicollinearity, you can:Drop one of the highly correlated variables (evaluate using Variance Inflation Factor, where $\text{VIF} > 5 \text{ or } 10$).Combine the correlated variables into a single feature (e.g., creating an average or index).Use dimensionality reduction techniques like Principal Component Analysis (PCA).Apply regularization methods such as Ridge or Lasso regression.

#13. What are some common techniques for transforming categorical variables for use in regression models?
-  Linear regression requires numerical inputs, so categorical variables must be transformed using:One-Hot Encoding (Dummy Encoding): Converts a categorical column into multiple binary (0 or 1) columns. To avoid the "dummy variable trap" (multicollinearity), one category must be dropped ($n-1$ dummies for $n$ categories).Ordinal Encoding: Assigns sequential integers to categories if they have a natural inherent order (e.g., Low = 1, Medium = 2, High = 3)
#14. What is the role of interaction terms in Multiple Linear Regression?
An interaction term is created by multiplying two independent variables together ($X_1 \times X_2$). Its role is to model scenarios where the effect of one predictor variable on the dependent variable depends on the level or magnitude of another predictor variable.
#15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?
-  Simple Regression: The intercept is simply the value of $Y$ when the single predictor $X$ is zero.Multiple Regression: The intercept is the expected value of $Y$ when all independent variables are simultaneously equal to zero. This value is often purely theoretical and may lack practical meaning if setting all predictors to zero is physically impossible in the real world (e.g., height, weight, age).
#16. What is the significance of the slope in regression analysis, and how does it affect predictions?
- The slope dictates both the direction and magnitude of the relationship between variables. In making predictions, the slope serves as a multiplier. If a feature's slope is large, minor adjustments to that input variable will generate substantial shifts in the model's final predicted value.
#17. How does the intercept in a regression model provide context for the relationship between variables?
-   The intercept anchors the regression line in space by establishing a baseline starting value. It informs us where the system stands before any external effects from the predictor variables are factored in.
#18. What are the limitations of using $R^{2}$ as a sole measure of model performance?
-   $R^2$ will never decrease when you add more independent variables, even if those variables are completely irrelevant noise. This can trick you into thinking a model is improving when it is actually overfitting.It does not indicate whether the coefficient estimates or predictions are biased.
#19. How would you interpret a large standard error for a regression coefficient?
- A large standard error indicates that the coefficient estimate is imprecise and highly variable across different samples. This usually means the variable is not a strong, reliable predictor in its current state, often resulting in a wider confidence interval and a larger, non-significant p-value.

#20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?
Identification: In a scatter plot of residuals vs. predicted values, heteroscedasticity manifests as a clear pattern—most commonly a "fan" or "cone" shape where the spread of points widens or narrows across the plot.

Importance: It must be addressed (via transformations like log transforms, or using Weighted Least Squares) because failing to do so breaks standard hypothesis testing, making your p-values and confidence intervals incorrect.

#22. Why is it important to scale variables in Multiple Linear Regression?

#23. What is polynomial regression?
-   Polynomial regression is a form of regression analysis in which the relationship between the independent variable X and the dependent variable Y is modeled as an nth-degree polynomial. Though it fits a non-linear curve to the data, it is still considered a form of linear regression because the parameters/coefficients (β) themselves are linear.

#24. How does polynomial regression differ from linear regression?
 -   Linear Regression: Fits a straight line or flat hyperplane through the data, assuming a constant rate of change.

Polynomial Regression: Generates a curved line by incorporating higher-power terms (like X 2 ,X 3 ), allowing it to map complex, fluctuating relationships.

#25. When is polynomial regression used?
 -  It is used when the relationship between the predictors and the target variable is visibly non-linear or curvilinear (e.g., growth rates that accelerate over time, or relationships that reach a peak and then decrease, like yield vs. temperature).

```python
#31
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import numpy as np

# Sample Data
X = np.array([[1], [2], [3], [4]])
y = np.array([2, 5, 10, 17])

# Step 1: Transform features to polynomial (e.g., degree=2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Step 2: Fit a Linear Regression model on the transformed features
model = LinearRegression()
model.fit(X_poly, y)

# Predictions
predictions = model.predict(X_poly)
```
